```python
import pandas as pd

url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2021/2021-04-20/netflix_titles.csv"
df = pd.read_csv(url)

df.head()
```

```python
# Step 2: Basic Info
df.info()
df.isnull().sum()
```

```python
# Step 3: Clean and Process (fixed version)
df['country'] = df['country'].fillna('Unknown')
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), format='mixed', errors='coerce')
df['year_added'] = df['date_added'].dt.year
```

```python
# Step 4: Insights
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x='type', data=df)
plt.title("TV Shows vs Movies")
plt.show()

df['country'].value_counts().head(10).plot(kind='bar', title='Top 10 Content-Producing Countries')
plt.show()
```
